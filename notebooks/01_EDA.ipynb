# 🛰️ SpoofSentinel-GNSS: Exploratory Data Analysis

This notebook explores the GNSS signal dataset to identify patterns distinguishing authentic signals from spoofed ones.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load data
train_path = '../data/train.csv'
if os.path.exists(train_path):
    df = pd.read_csv(train_path)
    print("Data loaded successfully.")
    display(df.head())
else:
    print(f"Please place your dataset at {train_path} to run this notebook.")
```

## 1. Class Distribution
Analyzing the balance between authentic and spoofed signals.

```python
if 'label' in df.columns or 'spoofed' in df.columns:
    target = 'label' if 'label' in df.columns else 'spoofed'
    sns.countplot(x=target, data=df)
    plt.title('Class Distribution (Authentic vs Spoofed)')
    plt.show()
```

## 2. Feature Distributions
Histograms of key metrics like CN0 and Doppler shift, colored by class.

```python
features_to_plot = ['CN0', 'Carrier_Doppler_hz', 'Pseudorange_m']
for feat in features_to_plot:
    if feat in df.columns:
        plt.figure(figsize=(10, 4))
        sns.histplot(data=df, x=feat, hue=target, kde=True, element='step')
        plt.title(f'Distribution of {feat}')
        plt.show()
```

## 3. Correlator Symmetry Analysis
Visualizing EC vs LC to detect signal distortion.

```python
if 'EC' in df.columns and 'LC' in df.columns:
    plt.figure(figsize=(8, 8))
    sns.scatterplot(data=df, x='EC', y='LC', hue=target, alpha=0.5)
    plt.plot([df['EC'].min(), df['EC'].max()], [df['EC'].min(), df['EC'].max()], 'r--')
    plt.title('Early Correlator vs Late Correlator (Symmetry)')
    plt.show()
```
